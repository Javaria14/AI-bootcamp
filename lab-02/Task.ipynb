```python
# Create a tuple
fruits = ("apple", "banana", "cherry", "orange")
print("Original tuple:", fruits)

# Length
print("Length:", len(fruits))

# Access items
print("First item:", fruits[0])
print("Last item:", fruits[-1])

# Slice a tuple
print("Slice:", fruits[1:3])

# Check if item exists
print("Is banana in tuple?", "banana" in fruits)

# Update tuple (convert to list)
temp = list(fruits)
temp[1] = "kiwi"
fruits = tuple(temp)
print("After update:", fruits)

# Add an item
fruits += ("mango",)
print("After adding:", fruits)

# Unpack a tuple
(first, second, *others) = fruits
print(first)
print(second)
print(others)

# Loop through the tuple
for item in fruits:
    print(item)
```

```python
# Create a set
fruits = {"apple", "banana", "cherry"}
print("Original set:", fruits)

# Length
print("Length:", len(fruits))

# Check if item exists
print("banana" in fruits)

# Add one item
fruits.add("orange")
print("After add:", fruits)

# Update with another set
tropical = {"mango", "pineapple"}
fruits.update(tropical)
print("After update:", fruits)

# Remove an item
fruits.remove("banana")
print("After remove:", fruits)

# Loop through the set
for item in fruits:
    print(item)

# Set operations
set1 = {"apple", "banana", "cherry"}
set2 = {"banana", "kiwi", "orange"}

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
print("Difference:", set1.difference(set2))
```

```python
# Create a dictionary
car = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print("Original dictionary:", car)

# Access values
print(car["brand"])
print(car.get("model"))

# Keys, values and items
print(car.keys())
print(car.values())
print(car.items())

# Check if key exists
print("year" in car)

# Update a value
car["year"] = 2024

# Add a new key-value pair
car["color"] = "Red"
print(car)

# Remove an item
car.pop("model")
print(car)

# Loop through dictionary
for key, value in car.items():
    print(key, ":", value)

# Copy dictionary
new_car = car.copy()
print(new_car)
```
